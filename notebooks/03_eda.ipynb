# Exploratory analysis
All statistics below are calculated from collected AMFI source data.

```python
import pandas as pd
d=pd.read_csv('../data/processed/monthly_flows.csv',parse_dates=['month']).set_index('month')
d[['equity_inflow','equity_outflow','net_flow','equity_aum']].describe().T.assign(skew=d[['equity_inflow','equity_outflow','net_flow','equity_aum']].skew(),kurtosis=d[['equity_inflow','equity_outflow','net_flow','equity_aum']].kurt())
```
